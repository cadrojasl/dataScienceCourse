```python
# se hace una consulta donde se muestra el nombre de la estación junto al dia mes y año en que se tomó la medida
result = ('''
select e."Nombre", t.dia, t.mes, t.anho
from fact_medidad as m
join (
	select id, dia, mes, anho
	from dim_tiempo
) as t
on m.id_tiempo = t.id
	join
	(
	select id, "Nombre"
	from dim_estacion
	) as e
	on m.id_estacion = e.id;
''')
with engine.connect() as conn:
    df_consulta = pd.read_sql(result, conn)
    # se cambia el nombre a las columnas
    df_consulta = df_consulta.rename(columns={'anho': 'año'})
    print(df_consulta)

#haga una nueva consulta
```

```python
#Consulta 2: Medidas por estación, fecha y contaminante
result_2 = ('''
SELECT 
    e."Nombre", 
    t.dia, t.mes, t.anho,
    p."PM10", p."PM2.5", p."NO2", p."OZONO"
FROM fact_medidad AS m
JOIN dim_tiempo AS t ON m.id_tiempo = t.id
JOIN dim_estacion AS e ON m.id_estacion = e.id
JOIN dim_polutante AS p ON m.id_polutante = p.id;
''')

#Consulta 3: Medidas por localidad, día festivo y concentración de CO
result_3 = ('''
SELECT 
    e."Localidad", 
    t.dia, t.mes, t.anho, t.festivo,
    p."CO"
FROM fact_medidad AS m
JOIN dim_tiempo AS t ON m.id_tiempo = t.id
JOIN dim_estacion AS e ON m.id_estacion = e.id
JOIN dim_polutante AS p ON m.id_polutante = p.id;
''')
```

```python
with engine.connect() as conn:
    df_consulta_2 = pd.read_sql(result_2, conn)
    print("\nConsulta: Medidas por estación, fecha y contaminante")
    print(df_consulta_2)

    df_consulta_3 = pd.read_sql(result_3, conn)
    print("\nConsulta: Medidas por localidad, día festivo y concentración de CO")
    print(df_consulta_3)
```

```python
result = ('''
SELECT t.hora, p."PM10", p."PM2.5", p."NO", p."NO2", p."NOX", p."CO", p."OZONO"
FROM fact_medidad AS m
JOIN dim_tiempo AS t ON m.id_tiempo = t.id
JOIN dim_polutante AS p ON m.id_polutante = p.id;
''')
with engine.connect() as conn:
    df_consulta = pd.read_sql(result, conn)
    # se cambia el nombre a las columnas
    df_consulta = df_consulta.rename(columns={'anho': 'año'})
```

```python
print(df_consulta.head())
print(df_consulta.dtypes)
print(df_consulta.isnull().sum())
```

```python
import matplotlib.pyplot as plt

contaminantes = ['PM10', 'PM2.5', 'NO', 'NO2', 'NOX', 'CO', 'OZONO']

# Crear figura con subplots
fig, axes = plt.subplots(nrows=len(contaminantes), ncols=1, figsize=(12, 3 * len(contaminantes)), sharex=True)

for i, contaminante in enumerate(contaminantes):
    ax = axes[i]
    
    # Boxplot por hora
    df_consulta.boxplot(column=contaminante, by='hora', ax=ax,
           flierprops={'markersize': 4, 'markerfacecolor': 'red'})
    ax.set_ylabel(f'{contaminante}')
    ax.set_title(f'{contaminante} por hora del día')

# Ajustar diseño
fig.suptitle('')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
